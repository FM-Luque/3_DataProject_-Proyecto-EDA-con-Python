#### 3️⃣ Cargar `customer-details.xlsx`

Este archivo contiene información demográfica de los clientes, repartida en **3 hojas** según el año de alta del cliente (2012, 2013, 2014).


```python
# Vemos primero qué hojas tiene el Excel
excel_clientes = pd.ExcelFile(r"customer-details.xlsx")
print("Hojas disponibles:", excel_clientes.sheet_names)
```

```python
# Cargamos cada hoja por separado
df_2012 = pd.read_excel(excel_clientes, sheet_name="2012")
df_2013 = pd.read_excel(excel_clientes, sheet_name="2013")
df_2014 = pd.read_excel(excel_clientes, sheet_name="2014")

print("2012:", df_2012.shape)
print("2013:", df_2013.shape)
print("2014:", df_2014.shape)
```

#### 4️⃣ Unir las 3 hojas de `customer-details.xlsx`

Como las 3 hojas tienen exactamente las mismas columnas y representan el mismo tipo de información (solo cambia el año de alta del cliente), las unimos verticalmente con `pd.concat()`.

Antes de unirlas, añadimos una columna `hoja_origen` para no perder la trazabilidad de qué hoja proviene cada registro (nos puede ser útil más adelante).


```python
# Añadimos una columna que identifique el año/hoja de origen antes de concatenar
df_2012["hoja_origen"] = "2012"
df_2013["hoja_origen"] = "2013"
df_2014["hoja_origen"] = "2014"

# Concatenamos verticalmente (misma estructura de columnas)
df_clientes = pd.concat([df_2012, df_2013, df_2014], ignore_index=True)

df_clientes.head(3)
```

#### Comprobación de la unión

```python
# La suma de filas debe coincidir con la suma de las 3 hojas por separado
print("Filas 2012 + 2013 + 2014:", len(df_2012) + len(df_2013) + len(df_2014))
print("Filas df_clientes:", len(df_clientes))
```

```python
# Verificamos el tamaño de df_clientes 
print("Tamaño final df_clientes:", df_clientes.shape)

```

```python
# Revisamos el df_clientes
df_clientes.info()
```

```python
# Comprobamos que no haya IDs de cliente duplicados entre hojas (cada cliente debería aparecer una sola vez)
print("Duplicados por ID:", df_clientes["ID"].duplicated().sum())
```

#### 5️⃣ Seleccionar columnas necesarias

Eliminamos la columna `Unnamed: 0` de ambos datasets, ya que es un índice residual de la exportación original y no aporta información para el análisis.


```python
# Comprobamos qué columnas tiene cada dataset antes de limpiar
print("Columnas df_bank:", df_bank.columns.tolist())
print("Columnas df_clientes:", df_clientes.columns.tolist())
```

```python
# Eliminamos la columna índice residual Unnamed: 0 en ambos datasets
df_bank = df_bank.drop(columns=["Unnamed: 0"])
df_clientes = df_clientes.drop(columns=["Unnamed: 0"])

print("Columnas df_bank:", df_bank.columns.tolist())
print("Columnas df_clientes:", df_clientes.columns.tolist())
```

#### 6️⃣ Renombrar columnas para facilitar su interpretación

La clave de unión entre ambos datasets se llama `id_` en `df_bank` y `ID` en `df_clientes`. Unificamos el nombre a `id_cliente` en los dos datasets para que el merge sea claro y explícito.


```python
df_bank = df_bank.rename(columns={"id_": "id_cliente"})
df_clientes = df_clientes.rename(columns={"ID": "id_cliente"})

print("Columnas df_bank:", df_bank.columns.tolist())
print("Columnas df_clientes:", df_clientes.columns.tolist())
```

#### 7️⃣ Unir `df_bank` con `df_clientes` (merge)

Unimos ambos datasets a través de la clave común `id_cliente`.

Usamos un merge **`how="left"`** tomando como base `df_bank` (los datos de la campaña), para conservar todos los registros de campaña aunque no encontremos información demográfica de algún cliente.


```python
print("Clientes únicos en df_bank:", df_bank["id_cliente"].nunique(), "/", len(df_bank))
print("Clientes únicos en df_clientes:", df_clientes["id_cliente"].nunique(), "/", len(df_clientes))
```

```python
df = df_bank.merge(df_clientes, on="id_cliente", how="left")

print("Shape df_bank:", df_bank.shape)
print("Shape df_clientes:", df_clientes.shape)
print("Shape df (resultado del merge):", df.shape)
```

Comprobación del merge

```python
# El número de filas del resultado coincide con el de df_bank (merge left), por lo que el merge no ha duplicado registros
print("Numero de filas en df:", len(df))
print("numero de filas en df_bank:", len(df_bank))
```

#### 8️⃣ Guardar el dataset integrado

Guardamos el resultado de esta fase como 01_datos_integrados para poder continuar con la **Fase 2 (Limpieza y transformación)** sin tener que repetir la carga y el merge cada vez.


```python
df.to_csv("01_datos_integrados.csv", index=False)

```
